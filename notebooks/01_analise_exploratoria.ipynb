#Importação da biblioteca pandas

```python
import pandas as pd
```

#Leitura de CSVs

```python
investors = pd.read_csv("data/raw/investors.csv")
investments = pd.read_csv("data/raw/investments.csv")
contributions = pd.read_csv("data/raw/contributions.csv")
transactions = pd.read_csv("data/raw/transactions.csv")
portfolio_history = pd.read_csv("data/raw/portfolio_history.csv")
```

#Análise e Tratamento de Dados

##Carregamento de dados

```python
print("Investors", investors.shape)
print("Investments", investments.shape)
print("Contributions", contributions.shape)
print("Transactions", transactions.shape)
print("Portfolio History", portfolio_history.shape)

```

## Tipos de dados


```python
print("=== INVESTORS ===")
print(investors.dtypes)

print("\n=== INVESTMENTS ===")
print(investments.dtypes)

print("\n=== CONTRIBUTIONS ===")
print(contributions.dtypes)

print("\n=== TRANSACTIONS ===")
print(transactions.dtypes)

print("\n=== PORTFOLIO HISTORY ===")
print(portfolio_history.dtypes)
```

##Mudança do tipo de data

```python
investors["registration_date"] = pd.to_datetime(investors["registration_date"])
contributions["contribution_date"] = pd.to_datetime(contributions["contribution_date"])
transactions["transaction_date"] = pd.to_datetime(transactions["transaction_date"])
portfolio_history["reference_date"] =pd.to_datetime(portfolio_history["reference_date"])

print("=== TIPOS APÓS A CONVERSÃO ===")
print(investors.dtypes)
print(contributions.dtypes)
print(transactions.dtypes)
print(portfolio_history.dtypes)
```

##Verificação de valores nulos

```python
print("=== VALORES NULOS ===")

print("\nInvestors:")
print(investors.isnull().sum())

print("\nInvestments:")
print(investments.isnull().sum())

print("\nContributions:")
print(contributions.isnull().sum())

print("\nTransactions:")
print(transactions.isnull().sum())

print("\nPortfolio History:")
print(portfolio_history.isnull().sum())
```

##Investigação de valores nulos

```python
print("Investimentos sem ticker")
print(investments[investments["ticker"].isnull()])
```

##Verificação de duplicidade

```python
print("=== DUPLICIDADES ===")

print("\nInvestors:")
print(investors.duplicated().sum())

print("\nInvestments:")
print(investments.duplicated().sum())

print("\nContributions:")
print(contributions.duplicated().sum())

print("\nTransactions:")
print(transactions.duplicated().sum())

print("\nPortfolio History:")
print(portfolio_history.duplicated().sum())

```

##Verificação de duplicidade de IDs

```python
print("=== DUPLICIDADES IDs ===")

print("\nInvestors:")
print(investors["investor_id"].duplicated().sum())

print("\nInvestments:")
print(investments["investment_id"].duplicated().sum())

print("\nContributions:")
print(contributions["contribution_id"].duplicated().sum())

print("\nTransactions:")
print(transactions["transaction_id"].duplicated().sum())

print("\nPortfolio History:")
print(portfolio_history["history_id"].duplicated().sum())
```

##Análise dos investidores

###Perfil

```python
print("=== PERFIL DOS INVESTIDORES ===")

print(investors["investor_profile"].value_counts())
```

###Percentual por perfil

```python
print("\n=== PERCENTUAL POR PERFIL ===")

print(investors["investor_profile"].value_counts(normalize=True) * 100)
```

###Idade dos investidores

```python
print("\n=== IDADE DOS INVESTIDORES ===")

print("Média: ", investors["age"].mean())
print("Mediana : ", investors["age"].median())
print("Mínima : ", investors["age"].min())
print("Máxima :", investors["age"].max())
```

###Distribuição por faixas etárias

```python
print("\n=== FAIXAS ETÁRIAS ===")

faixas_etarias = pd.cut(
    investors["age"],
    bins=[17, 25, 35, 45, 55, 70],
    labels = ["18-25", "26-35", "36-45", "46-55", "56-70"]
)

print(faixas_etarias.value_counts().sort_index())
```

###Análise da Renda

```python
print("\n=== RENDA DOS INVESTIDORES ===")

print("Media: ", investors["income"].mean())
print("Mediana: ", investors["income"].median())
print("Mínima: ", investors["income"].min())
print("Máxima: ", investors["income"].max())
```

###Renda por perfil de investidor

```python
print("\n=== RENDA POR PERFIL ===")

print(
    investors.groupby("investor_profile")["income"].mean()
)
```

##Visão geral das contribuições

###Análise das contribuições

```python
print("\n=== CONTRIBUIÇÕES ===")

print("Quantidade: ", contributions["contribution_id"].count())
print("Total: ", contributions["amount"].sum())
print("Média: ", contributions["amount"].mean())
print("Mediana: ", contributions["amount"].median)
print("Mínimo: ", contributions["amount"].min())
print("Máximo: ", contributions["amount"].max())
```

###Evolução das contribuições ao longo do tempo

```python
print("\n=== CONTRIBUIÇÕES POR MÊS ===")

contribuicoes_mensais = (
    contributions
    .groupby(contributions["contribution_date"].dt.to_period("M"))["amount"]
    .sum()

)

print(contribuicoes_mensais)
```

###Contribuições por investidor

```python
print("\n=== CONTRIBUIÇÕES POR INVESTIDOR ===")

contribuicoes_por_investidor = (
    contributions
    .groupby("investor_id")["amount"]
    .sum()
)

print(contribuicoes_por_investidor.head())
```

###Contribuições por perfil

```python
print("\n=== CONTRIBUIÇÕES POR PERFIL ===")

contribuicoes_perfil = (
    contributions
    .merge(investors[["investor_id", "investor_profile"]], on="investor_id")
    .groupby("investor_profile")["amount"]
    .sum()
)

print(contribuicoes_perfil)
```
